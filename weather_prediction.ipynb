```python
t=int(input('Enter Temperature '))
h=int(input('Enter Humidity '))
wind=int(input('Enter Wind Speed '))
w=0.5*(t**2)-(0.2*h)+(0.1*wind)-15
if w>10:
  print("Sunny")
elif 0<=w and w<=10:
  print("Windy")
elif -10<=w and w<0:
  print("Cloud")
elif w<-10:
  print("Rainy")
```
